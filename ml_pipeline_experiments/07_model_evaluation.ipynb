```python
import os
import joblib
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import r2_score, mean_absolute_error, root_mean_squared_error

# 1. Thiết lập đường dẫn tương đối từ thư mục 'ml_pipeline_experiments'
data_dir = os.path.join('..', 'data', 'processed')
model_dir = os.path.join('..', 'backend', 'models')

# 2. Tải dữ liệu tập Test
test_csv_path = os.path.join(data_dir, 'test.csv')
test_df = pd.read_csv(test_csv_path)

# 3. Tách X_test, y_test
X_test = test_df.drop(columns=['price_log']).values
y_test = test_df['price_log'].values

print(f"-> Đã tải thành công dữ liệu Test từ: {test_csv_path}")
print(f"-> Kích thước tập Test: X_test {X_test.shape} | y_test {y_test.shape}")
```

```python
# Tải mô hình đã được huấn luyện và chọn lọc từ file 06
model_path = os.path.join(model_dir, 'best_stacking_regressor.pkl')
best_model = joblib.load(model_path)

print(f"-> Đã tải thành công mô hình từ: {model_path}")
print("-> Kiến trúc mô hình:")
print(best_model)
```

```python
# Thực hiện dự đoán trên dữ liệu mù (unseen data)
preds_test = best_model.predict(X_test)

# Tính toán các độ đo
test_r2 = r2_score(y_test, preds_test)
test_mae = mean_absolute_error(y_test, preds_test)
test_rmse = root_mean_squared_error(y_test, preds_test)

# In Report
print("="*55)
print("  BÁO CÁO HIỆU NĂNG MÔ HÌNH TRÊN TẬP TEST (HOLD-OUT)")
print("="*55)
print(f"  - R2 Score (Độ giải thích):      {test_r2:.4f}")
print(f"  - MAE (Sai số tuyệt đối):        {test_mae:.4f}")
print(f"  - RMSE (Sai số căn bình phương): {test_rmse:.4f}")
print("="*55)
print("=> LƯU Ý: Đây là các con số cuối cùng mang tính khách quan tuyệt đối để đưa vào CV!")
```
