```python
import pandas as pd
import os

input_dir = "../data/raw"
input_path = os.path.join(input_dir, "01_text_parsing_and_feature_enhancement.csv")

if not os.path.exists(input_path):
    raise FileNotFoundError(f"Không tìm thấy file: {input_path}. Hãy chạy lại file 01 trước.")

df = pd.read_csv(input_path)
print(f" Đã nạp thành công bộ dữ liệu nền. Kích thước: {df.shape[0]} dòng.")

# Tập 1: Đất nền / Đất thổ cư 
df_land = df[df['title'] == 'Đất nền / Đất thổ cư'].reset_index(drop=True)

# Tập 2: Nhà riêng / Nhà phố
df_house = df[df['title'] == 'Nhà riêng'].reset_index(drop=True)

# Tập 3: Chung cư / Căn hộ
df_apartment = df[df['title'] == 'Chung cư / Căn hộ'].reset_index(drop=True)

output_dir = "../data/raw"
os.makedirs(output_dir, exist_ok=True)

path_land = os.path.join(output_dir, "02_properties_land_raw.csv")
path_house = os.path.join(output_dir, "02_properties_house_raw.csv")
path_apartment = os.path.join(output_dir, "02_properties_apartment_raw.csv")

df_land.to_csv(path_land, index=False)
df_house.to_csv(path_house, index=False)
df_apartment.to_csv(path_apartment, index=False)

print("\n")
print("  QUY TRÌNH PHÂN TÁCH DỮ LIỆU THÀNH CÔNG (LƯU TẠI RAW):")
print("\n")
print(f" 1. Tập ĐẤT NỀN THÔ       : {df_land.shape[0]} dòng.")
print(f"    -> Lưu tại: {path_land}")
print(f" 2. Tập NHÀ RIÊNG THÔ     : {df_house.shape[0]} dòng.")
print(f"    -> Lưu tại: {path_house}")
print(f" 3. Tập CHUNG CƯ CĂN HỘ THÔ: {df_apartment.shape[0]} dòng.")
print(f"    -> Lưu tại: {path_apartment}")
print("\n")
total_split = df_land.shape[0] + df_house.shape[0] + df_apartment.shape[0]
print(f" Tổng số dòng sau phân tách: {total_split} dòng (Khớp 100% với dữ liệu gốc).")
```
